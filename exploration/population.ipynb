```python

df = spark.table('dev.my_mlops_project.population')
df_pf = spark.table('dev.my_mlops_project.population_features')

df.printSchema()
df_pf.printSchema()
```

```python
df = df_pf

n_rows = df.count()
n_year = df.select('year').distinct().count()
year_start = df.select("year").orderBy("year").first()["year"]
year_end = df.select("year").orderBy("year", ascending=False).first()["year"]
print(f"Number of rows: {n_rows}")
print(f"Number of years: {n_year}")
print(f"Year start: {year_start}")
print(f"Year end: {year_end}")

```

```python

for col in df.columns:
    print(f"Number of null values in column {col}: {df[df[col].isNull()].count()}")
```
